```python
from typing import TypedDict
from langgraph.graph import StateGraph

class AgentState(TypedDict):
    AImessage : str
    HUMANmessage : str

def askname(state : AgentState) -> AgentState:
    state["message"] = input("What's your name you rat?")
    return state


def greetingnode(state : AgentState) -> AgentState :
    state["message"] = "It's John Marston " + state["message"] + "!!!!"
    return state


```

```python
graph = StateGraph(AgentState)

graph.add_node("askname",askname)
graph.add_node("greetingnode",greetingnode)

graph.add_edge("askname","greetingnode")

graph.set_entry_point("askname")
graph.set_finish_point("greetingnode")

app = graph.compile()
```

```python
print(app.get_graph().draw_mermaid())
```

```python
result = app.invoke({"message": ""})
print(result)
```
